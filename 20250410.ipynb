수평으로 던진 공의 운동

```python
#초기속도 입력
v0 = float(input())
print("초기속도 :", v0, "m/s")
```

```python
import math
#초기 각도 입력
angle_degree = float(input())
print("초기 각도 :", angle_degree, "도")

#각도를 radian으로 바꾸기
angle_radian = math.radians(angle_degree)
print("초기 각도 :", angle_radian, "ranian")
```

```python
#x성분의 속도를 구하세요
v0x = math.cos(angle_radian) * v0
print("x축의 초기속도 :", v0x, "m/s")
```

```python
#y성분의 속도를 구하세요
v0y = math.sin(angle_radian) * v0
print("y축의 초기속도 :", v0y, "m/s")
```

```python
#최고점의 높이
g = 9.8#m/s^2
t= v0y / g
a = -g #m/s^2
H = g * t * t / 2
print("최고점의 높이 :", H, "m")
```

```python
#최고점 도착 시간
t= v0y / g
print("최고점 도착 시간 :", t, "s")
```

```python
#수평으로 이동한 최고거리
x = v0x * t * 2
print("수평으로 이동한 최고거리:", x, "m")
```

```python
#지면에 닿을때의 Vx, Vy를 구하고 V를 벡터 성분을 구하세요
Vx = v0 * math.cos(angle_radian)
Vy = v0 * math.sin(angle_radian)
V = math.sqrt(Vx**2 + Vy**2)
theta = math. atan(Vy/Vx)
angle = math.degrees(theta)
print("지면에 닿을때 Vx :", Vx, "m/s")
print("지면에 닿을때 Vy :", Vy, "m/s")
print("지면에 닿을때 V 백터 성분 :", V,"m/s",  angle, "°")
```
